The basic structure is:

while condition:
    # code

water_level = 0

while water_level < 100:
    water_level = water_level + 10

Here:

water_level = 0 → initialization
water_level < 100 → condition
water_level += 10 → update

If we forget the update?

We can accidentally create an infinite loop.

For example:

number = 1

while number <= 5:
    print(number)

This never changes number.

So Python keeps printing 1.

That's why every important while loop should clearly have:

Initialization
Condition
Update / Termination

## When should we use while instead of for?

A for loop is usually useful when we know what we're iterating over 
or approximately how many iterations we need.

```python
#Example
for number in range(1, 101): 
    print(number)
```

Known sequence/count → for loop
Unknown sequence/count → while loop

```python
#Print numbers from 1 to 100

number = 1 #initialization

while number<= 100: #condition
    print(number)
    number = number +1 #update /Termination
```

```python
#print number from 100 to 1

number = 100 #initialize

while number >=1:#condition
    print(number)
    number = number -1 #update
```

```python
#print even numbers from 1 to 100

number = 1 #initilize

while number <=100:#condition
    if number % 2 == 0:
        print(number)
    number = number +1 #update
```

```python
#calculate the sum of digits

#5832 -- 5+8+3+2 = 18

#initialize
number = 5832
total = 0

while number > 0:
    digit = number % 10 # 2 , 3
    total = total + digit # 0+2 = 2, 2+3=5 
    number //= 10 # Remove last digit 2 , new number = 583, remove 3 nuew number 58
print(total)
```

```python
#reverse an integer

# 12345 - 54321

#initialize
number = 12345
reverse = 0

while number >0: #condition
    digit = number % 10 # Gives last digit
    reverse = reverse*10 + digit # (0+5)= 5, (5*10+4 = 54)
    number //= 10 #update 1234
print(reverse)
```

```python
#Count the digits
#initialization
number = 5832 #4digits
count = 0

#condition
while number > 0:
    number= number // 10
    count = count +1 #update
print(count)
```

```python
#Calculate factorial using while

#5! = 5*4*3*2*1
#120

number = 5

# Initialization
factorial = 1

while number > 0:  # Condition
    factorial *= number
    number -= 1  # Update / Termination

print("Factorial:", factorial)
```

```python
#Create a program that repeatedly asks for numbers and stops only when the user enters 0.
#  Display the sum of all previously entered numbers.

total = 0

number = int(input("Enter a number (0 to stop): "))

while number != 0:  # Condition
    total += number
    number = int(input("Enter a number (0 to stop): "))  # Update / Termination

print("Total:", total)
```

```python
#create password checker

#intialize

correct_password = "python123"
password = ""

while password!= correct_password: #condition
    password = input("Enter password: ") #update 

print("Login Sucessful")
```

```python
#Guessing game
#secret number = 42

secret_number = 42 #initialize
guess = None

while secret_number != guess: #condition
    guess= int(input("Guess the number: ")) #update statment

print("Correct!,You guessed the number")
```

```python
# 11. Menu-Driven Calculator

while True:
    # Display the menu
    print("\n--- Calculator ---")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")

    # Take user's choice
    choice = input("Enter your choice: ")

    # Termination
    if choice == "5":
        print("Exiting calculator...")
        break

    # Check whether choice is valid
    if choice in ("1", "2", "3", "4"):

        # Take two numbers
        first = float(input("Enter first number: "))
        second = float(input("Enter second number: "))

        # Perform operation
        if choice == "1":
            print("Result:", first + second)

        elif choice == "2":
            print("Result:", first - second)

        elif choice == "3":
            print("Result:", first * second)

        elif choice == "4":
            if second == 0:
                print("Cannot divide by zero.")
            else:
                print("Result:", first / second)

    else:
        print("Invalid choice. Please select 1 to 5.")
```

```python
# Initialization
balance = 10000

while True:  # Condition: Continue until user chooses Exit

    print("\n--- ATM MENU ---")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("Current Balance:", balance)

    elif choice == "2":

        amount = float(input("Enter deposit amount: "))

        if amount > 0:
            balance += amount
            print("Deposit successful.")
            print("New Balance:", balance)
        else:
            print("Invalid amount.")

    elif choice == "3":

        amount = float(input("Enter withdrawal amount: "))

        if amount <= 0:
            print("Invalid amount.")

        elif amount > balance:
            print("Insufficient balance.")

        else:
            balance -= amount
            print("Withdrawal successful.")
            print("Remaining Balance:", balance)

    elif choice == "4":
        break  # Termination: User selected Exit

    else:
        print("Invalid choice.")
```
