```python
# 1 ) Transaction Total Without sum()

transactions = [1200,450,800,1500,2300,700,100]

total = 0

for transaction in transactions:
    total +=transaction

print(total)
```

```python
#2) Find Highest and Lowest Transaction

#max() min()
transactions = [1200,450,800,1500,2300,700,100]

highest= transactions[0]
lowest = transactions[0]


for transaction in transactions:
    if transaction > highest:
        highest = transaction

    if transaction < lowest:
        lowest = transaction

print("Highest transaction:",highest)
print("Lowest transaction:",lowest)
```

```python
#3) Calculate Average Temperature

temperatures = [32,35,28,40,38,31,42]

#average = total / number of values

total = 0
count = 0

for temprature in temperatures:
    total += temprature
    count += 1

average = total /count

print("Average temprature: ", average)
```

```python
#4) Student Marks Categorization .

marks = [78,92,45,67,88,53,99]

# 90+ , 75-89, 50- 74, below 50

above_90 = 0

between_75_89 = 0
between_50_74 = 0
below_50 =0

for mark in marks:

    if mark >90:
        above_90 += 1
    elif mark>=75:
        between_75_89 += 1
    elif mark>=50:
        between_50_74 += 1
    else:
        below_50 +=1

print("90+", above_90)
print("75-89", between_75_89 )
print("50-74", between_50_74)
print("Below_50", below_50)
```

```python
#8) Character Frequency Without Counter

text = "banana"

frequency = {}

for character in text:
    if character in frequency:
        frequency[character] += 1
    else:
        frequency[character] = 1

for character,count in frequency.items():
    print(character,"->", count)
```

```python
#9) Find the Second-Largest Number Without sort()

numbers = [10,25,7,40,32,18]

largest = float("-inf")
second_largest = float("-inf")

for number in numbers:

    if number >largest:
        second_largest = largest
        largest = number

    elif number>second_largest and number!= largest:
        second_largest = number

print("largest:",largest)
print("Second largest",second_largest)
```

```python
#1. Check Balance
def check_balance(balance):
    print(f"Current balance: ₹{balance}")
```

```python
def deposit(balance):
    amount = float(input("Enter deposit amount: ₹"))

    if amount > 0:
        balance += amount
        print(f"₹{amount} deposited successfully.")
    else:
        print("Invalid deposit amount.")

    return balance
```

```python
def withdraw(balance):
    amount = float(input("Enter withdrawal amount: ₹"))

    if amount <= 0:
        print("Invalid withdrawal amount.")

    elif amount > balance:
        print("Insufficient balance.")

    else:
        balance -= amount
        print(f"₹{amount} withdrawn successfully.")

    return balance
```

```python
def atm():

    balance = 10000

    while True:

        print("\n===== ATM MENU =====")
        print("1. Check Balance")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            check_balance(balance)

        elif choice == "2":
            balance = deposit(balance)

        elif choice == "3":
            balance = withdraw(balance)

        elif choice == "4":
            print("Thank you for using the ATM.")
            break

        else:
            print("Invalid choice. Please try again.")


```
