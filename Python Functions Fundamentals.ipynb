```python
#basic calculator using functions


#addition
def add(a,b):
    return a+b

#substraction

def subtract(a,b):
    return a-b

#multiplication
def multiply(a,b):
    return a*b

#division
def division(a,b):

    if b ==0:
        return "Cannot divide by zero"
    else:
        return a/b
```

```python
print("Addition",add(10,5))
print("Substraction",subtract(10,5))
print("Multiplication",multiply(10,5))
print("Division",division(10,5))
```

```python
print("Division",division(10,0))
```

```python
#Number is even or odd

def check_even_odd(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"
```

```python
print(check_even_odd(10))
print(check_even_odd(7))
```

```python
#largest of 3 number

def largest_of_three(a,b,c):
    largest = a

    if b> largest:
        largest = b
    elif c>largest:
        largest = c
    return largest
```

```python
print(largest_of_three(24,50,40))
```

```python
#factorial function

#5 factorial = 5*4*3*2*1

def factorial(number):

    result = 1

    for i in range(1,number+1):
        result *= i
    return result
```

```python
factorial(5)
```

```python
#prime number

# eg 2,3,5,7,11

def is_prime(number):

    if number<2:
        return False

    for i in range(2,number):
        if number % i == 0:
            return False

    return True
```

```python
is_prime(7)
```

```python
is_prime(10)
```

```python
#Default arguments

def calculate_discont(price, discount = 10):

    discount_amount = price*discount/100

    final_price = price - discount_amount

    return final_price


```

```python
print(calculate_discont(1000))
```

```python
print(calculate_discont(1000,20))
```

```python
#function to calculate list sum

def calculate_total(numbers):

    total = 0

    for number in numbers:
        total += number

    return total
```

```python
calculate_total([100,200,300,400])
```

```python
#count vowels

#aeiou

def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for character in text:
        if character in vowels:
            count += 1

    return count
```

```python
print(count_vowels("Python Programming"))
```

```python
#palindrome function

#eg - madam

def is_palindrome(text):
    reversed_text = ""

    for character in text:
        reversed_text = character +reversed_text

    return text == reversed_text
```

```python
print(is_palindrome("Madam"))
```

```python
print(is_palindrome("madam"))
```
