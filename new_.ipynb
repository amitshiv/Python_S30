```python
#print numbers from 1 to 100 but skip number divisiblke by 5
#continue

for number in range(1,101):

    if number % 5 == 0:
        continue
    print(number)

```

```python
#Break statment

for number in range(1,101):
    if number % 7 == 0 and number % 11 == 0:
        print("First matching number",number)
        break
```

```python
#Remember
#Continue = Skip the iteration
#break = stop the entire loop
```

```python
#for - else

numbers = [10,25,30,45,50,75]

search_number = int(input("Enter a number to serach:"))

for number in numbers:
    if number == search_number:
        print("Number Found")
        break
else:
    print("Number not found")
```

```python
#enumerate

name = ["Aman", "Ravi","Sudh","priya","Anjali"]

for i,j in enumerate(name,start = 1):
    print(i,j)
```

```python
#Generate multiplication tables from 1 to 10 using nested loops.

for table in range(1,11):
    print(f"\n Mulitiplication Table of {table}")

    for number in range(1,11):
        result = table * number
        print(f"{table} X {number} = {result}")
```

```python
#Given a list containing duplicate elements, create another list containing only unique elements without using set().


numbers = [10,20,10,30,20,40,30,50]

unique_numbers = []

for number in numbers:

    if number not in unique_numbers:
        unique_numbers.append(number)

print("unique_numbers", unique_numbers)
```

positive numbers

```python
numbers = [10, -4, 8, -2, 0, 15, -9, 21]

positive_count = 0
negative_count = 0
zero_count = 0

for number in numbers:

    if number >0:
        positive_count = positive_count+1

    elif number <0:
        negative_count = negative_count +1

    else:
        zero_count = zero_count +1

print("positive_count",positive_count)
print("negative_count",negative_count)
print("zero_count",zero_count)
```

```python
#Write a program to determine whether a number is prime using a loop.
```

```python
number = int(input("Enter a number: "))

if number<2:
    print("Not a prime number")
else:
    is_prime = True

    for i in range(2,number):

        if number % i == 0:
            is_prime = False
            break
    if is_prime:
        print("Prime number")
    else:
        print("Not a prime number")
```

```python
#Print all prime numbers between 1 and 100.

for number in range(2,101):

    is_prime = True

    for i in range(2,number):
        if number % i == 0:
            is_prime = False
            break
    if is_prime:
        print(number)

```
